```python
import numpy as np
```

```python
rows=int(input("No of rows:"))
cols=int(input("No of columns:"))
A=np.zeros((rows,cols),dtype=int)
for i in range(rows):
    for j in range(cols):
        A[i][j]=int(input("Enter element:"))
transpose=np.zeros((cols,rows),dtype=A.dtype)
for i in range(rows):
    for j in range(cols):
        transpose[j][i]=A[i][j]
print("Original Matrix")
print(A)
print("\nTranspose of A")
print(transpose)
```
